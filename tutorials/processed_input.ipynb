# Processed input example
This notebook creates a small artificial two-view input and runs the processed-input interface. It is only a format demonstration; it is not a biological dataset or a paper benchmark.

```python
from pathlib import Path
import subprocess
import sys

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').is_file())
example = root / 'example.npz'
subprocess.run([sys.executable, str(root / 'scripts/make_example.py'), '--output', str(example)], check=True)
subprocess.run([sys.executable, '-m', 'smahd.cli', '--input', str(example), '--output', str(root / 'output/example'), '--device', 'cpu', '--full-graph', '--epochs', '2'], check=True)
```
